```python
df[df['Duration'] <= 51] 
```

#### USING WHERE

```python
# where filter the rows and coiloum will null /nun
    # instead of removing it completly

df.where(df['Duration'] <= 51)
```

```python
# if you want to use your own message or somethig
    # instead of null/nun

# use         (   other     )  

df.where(df['Duration'] <= 51,other='not eligable')
```
