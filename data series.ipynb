# DATA SERIES

```python
import pandas as pd
import numpy as np
```

```python
# Data series are much similar as python list

# Data series are ordered 

# --------------------------------------------------
ds=pd.Series([i for i in range(1,100)])
ds
```

## selecting from a series

```python
#
# Filtering / Boolean Masking
print(ds[ds > 2])      # Returns elements greater than 2 (3 and 4)#
```

```python
# #Slicing
print(ds[2:6])  
```

```python
# Position-based selection
print(ds.iloc[1])    
```

## Handling Missing Data

```python
print(ds.isna())      # Returns a Boolean Series checking for missing values
```

```python
print(ds.fillna(0))   # Replaces missing values with 0
```

```python
print(ds.dropna())    # Removes missing values
```

##  Vectorized Operations

```python
# Element-wise math
print(ds * 10)        # Multiplies every element by 10
```

```python
# Filtering / Boolean Masking
print(ds[ds > 2])      # Returns elements greater than 2 (3 and 4)
```
