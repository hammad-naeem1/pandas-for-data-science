#### concatenate and merge

```python
import pandas as pd
```

```python
df1=pd.DataFrame({'ID':[1,2,3],'Score':[41,58,69]})
df1 = df1.set_index('ID')
df1
```

```python
df=pd.read_csv('data.csv')
df
```

```python
df2=pd.read_csv('data.csv')
df2
```

#### mergeing  join

```python
pd.merge(df2,df,how='inner',right_on='Calories',left_on='Calories')
```

#### concat 

```python
table=pd.concat([df1,df])
table
```

####  vertically join or row wise

```python
new_table=pd.concat([df,df1],axis=0)
new_table
```

####  horizintally  join or coloumn wise

```python
new_table=pd.concat([df,df1],axis=1)
new_table
```
