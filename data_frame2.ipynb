##### basic dataframe understanding

```python
import pandas as pd
```

```python
# reading a csv file 
df=pd.read_csv('data.csv')
df
```

```python
# using head()  gives us top 5 tables by default 
# but we can get as much elements we want! but giving our own num as parametre

df.head()
df.head(10)
```

```python
# using tail() gives us  bottom 5 tables by default 
# but we can get as much elements we want! but giving our own num as parametre

df.tail()
df.tail(10)
```

```python
# we can use shape like we use in numpy
# to determine the rows and coloums
# ---------------------------------
df.shape
```

```python
# we can use columns to look up the columns

df.columns
```

```python
# too rename a column we use rename

df.rename(columns={ 'Subscription Date':'Subscription ending date'},inplace=True)
# ------------
# we can use  (  inplace = True  )  to save our change in the column
```

```python
# too get information about data_frame we can use  -> info()
df.info()
```

```python
# we can use describe also to show case statical summary about dataframe
df.describe()
```
