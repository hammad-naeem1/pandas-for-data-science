#### aggregation and group by

```python
df=pd.read_csv('data.csv')
df
```

```python
df['Pulse'].value_counts()
```

### aggregation based on group by

```python
df.groupby('Calories')['Pulse'].sum()
```

#### us9ng group by multiple aggregatetion

```python
df.groupby('Calories').agg({'Pulse' :'mean' ,'Date':'count'})
```
